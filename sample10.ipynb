```python
import pandas as pd
import re

# Load data
df = pd.read_csv("final_final_video_comments.csv", quotechar='"')

# Check total rows
print(f"Total rows: {len(df)}")

# Function to repair misaligned columns from index 10000 to 10433
def repair_misaligned_rows(df, start_index=9935, end_index=10433):
    for i in range(start_index, end_index + 1):  # Adjusted to stop at 10433
        row = df.iloc[i]
        video_id = row["video_id"]
        
        # Ensure we are dealing with misaligned rows where video_id is misplaced
        if isinstance(video_id, str) and ',' in video_id:
            # Using regex to handle splitting by commas but avoiding commas inside quotes
            parts = re.split(r',(?=(?:[^\"]*\"[^\"]*\")*[^\"]*$)', video_id)
            
            # Check if we have 3 parts, then reassign to the correct columns
            if len(parts) == 3:
                # Reassign the values to their respective columns
                df.at[i, "video_id"] = parts[0]
                df.at[i, "video_title"] = parts[1]
                df.at[i, "comment"] = parts[2]

    return df

# Repair rows from index 10000 to 10433
df = repair_misaligned_rows(df)
df = repair_misaligned_rows(df,start_index=14214, end_index=14682)

# Check first 10 rows after fixing
display(df.head(10))

# Optionally save the corrected file
# df.to_csv("corrected_video_comments.csv", index=False)
```

```python
import re  # Add this at the top of your script for regex processing

def normalize_text(text):
    """Removes special symbols and extra spaces while preserving important content."""
    text = re.sub(r"[^\w\s,]", "", text)  # Remove non-word characters except spaces and commas
    text = re.sub(r"\s+", " ", text).strip()  # Normalize extra spaces
    return text
```

```python
import requests
import json
import pandas as pd
import time
import os

# === CONFIGURATION ===
df_sample = df.head(12000).copy()  # Sample data
OUTPUT_FILE = "processed_comments.csv"  # Output file
BATCH_SIZE = 100  # Save progress every BATCH_SIZE comments
API_URL = "http://localhost:1234/v1/completions"  # LM Studio API endpoint
MODEL_NAME = "mistral-7b-instruct@q3_k_l"  # Model in LM Studio
MAX_RETRIES = 4  # Number of retries for API errors
PROMPT_DELAY = 0.3  # Delay between API requests

# === LOAD DATA & RESUME IF NECESSARY ===
try:
    processed_df = pd.read_csv(OUTPUT_FILE)  
    last_index = len(processed_df) - 1  # Get the last processed index
    print(f"Resuming from index {last_index + 1}...")

    # Ensure df_sample has the same columns as the saved file
    df_sample["label"] = ""

    # Only copy existing processed data if last_index is valid (i.e., there is existing data)
    if last_index >= 0:
        df_sample.loc[:last_index, ["comment", "label"]] = processed_df.loc[:last_index, ["comment", "label"]]

except FileNotFoundError:
    df_sample["label"] = ""  # If no file exists, start fresh
    last_index = -1  


# === FUNCTION TO HANDLE API REQUESTS WITH RETRIES ===
def correct_spelling_deepseek(text):
    """Calls LM Studio API to correct spelling and label the comment."""
    prompt = f"""
You are a data labeler and a spell checker with the main task of interpreting social media comments on trending YouTube Shorts videos from Indonesia and assigning a label to each of them. 
Your task is to spellcheck and interpret the meaning of the social media comments. 
Label the comment as either "Online Gambling Promotion" if it promotes online gambling, or "No Online Gambling Promotion" otherwise.

Correct the spelling in this Indonesian text if it is not understandable. Preserve gambling terms like 'judi', 'slot', 'bonus', 
'gacor', 'maxwin', or any other keywords that might indicate online gambling promotions. 

Output strictly in this format: <corrected_comment>, <label> 
Do not add punctuations to the comments only add them to differentiate comment and label

Don't give any explanations—just output the corrected or uncorrected comment along with its label. Make sure the output follows this format:
Corrected comment without parentheses: "<corrected_comment>, <label>"

Example of Expected Format:
1. sy mau daptar judi onlen -> "saya mau daftar judi online, Online Gambling Promotion"
2. bersandar teras rumah sembari ngopi bermain permainan nyelot jptogel77 nikmat sekaleh → 
    "bersandar teras rumah sembari ngopi bermain permainan slot jptogel77 nikmat sekali, Online Gambling Promotion"
3. lagi nunggu ojol di pinggir jalan →  "lagi nunggu ojol di pinggir jalan, No Online Gambling Promotion"

Output format enforcement: **Do NOT include explanations or extra words. Only return the formatted result.**  
DO NOT put any word like "Output:" or "Answer:" before the corrected_comment

Text to process: "{text}"
"""

    payload = {
        "model": MODEL_NAME,
        "prompt": prompt,
        "max_tokens": 70,
        "temperature": 0.3
    }

    headers = {"Content-Type": "application/json"}

    for attempt in range(MAX_RETRIES):
        try:
            response = requests.post(API_URL, headers=headers, json=payload)
            response_json = response.json()
            
            # Extract the generated response
            corrected_text = response_json["choices"][0]["text"].strip()

            # Clean the text by removing unwanted prefixes (Label, Output, Answer) until punctuation
            unwanted_prefixes = ['label', 'output', 'answer']
            while any(corrected_text.lower().startswith(prefix) for prefix in unwanted_prefixes):
                # Find the first punctuation mark
                for punct in [ ':']:
                    punct_index = corrected_text.find(punct)
                    if punct_index != -1:
                        corrected_text = corrected_text[punct_index + 1:].strip()
                        break
                if not corrected_text:  # If no punctuation is found, break the loop
                    break

            return corrected_text if corrected_text else text  

        except Exception as e:
            print(f"Error on attempt {attempt+1}: {e}")
            time.sleep(1 * attempt)  # Exponential backoff (2s, 4s, 8s...)

    return text  

# === PROCESS COMMENTS (1001 to 2000) ===
# start_index = max(9936, last_index + 1)  # Start from the 1001st comment or after the last processed index
start_index = 9936  # Start from the 1001st comment or after the last processed index
end_index = min(9940, len(df_sample))   # End at the 2000th comment or at the last comment

# Debugging prints to check range and last_index
print(f"Last processed index: {last_index}")
print(f"Processing comments from index {start_index} to {end_index}")

# Initialize a list to store processed data (not modifying df_sample directly)
processed_comments = []

# Check if there are comments to process
if start_index >= end_index:
    print("No comments to process in the specified range.")
else:
    # Process from start_index to end_index (but don't send to API, just print)
    for i in range(start_index, end_index):
        video_id = df_sample.loc[i, "video_id"]
        video_title = df_sample.loc[i, "video_title"]
        comment = df_sample.loc[i, "comment"]
        
        # Prepare and send the original comment to the API
        processed_text = correct_spelling_deepseek(comment).replace("\\", "").replace('"', "").replace('\n', "")
        
        print(f"Sending comment to API (index {i}): {comment}")
        print(f"Processed text response: {processed_text}")
        
        # After the API returns, we process the response
        retry_attempts = 0  # Track retry attempts
        max_retries = 5  # Max retries for ValueError
        
        while retry_attempts < max_retries:
            try:
                processed_text = processed_text.strip('"')  # Normalize text
                print(f"processed_text : {processed_text}")
                
                # Correct the splitting issue: now, split the text into corrected comment and label correctly
                # We split at the first comma, which separates the comment from the label
                corrected_comment, label = processed_text.split(",", 1)
                corrected_comment = corrected_comment.strip()
                label = label.strip()
                break  # Exit loop if successful
            except ValueError:
                print(f"ValueError encountered. Retrying... (Attempt {retry_attempts + 1})")
                processed_text = correct_spelling_deepseek(comment).replace("\\", "").replace('"', "").replace('\n', "")
                retry_attempts += 1
                time.sleep(PROMPT_DELAY)

        if retry_attempts == max_retries:  
            print("Max retries reached. Assigning default values.")
            corrected_comment, label = processed_text.strip(), "ERROR"

        # Store the processed comments and labels in a separate list (without modifying df_sample)
        processed_comments.append({
            'video_id': video_id,
            'video_title': video_title,
            'comment': comment,
            'label': label
        })

        # Save progress every BATCH_SIZE comments
        if (i + 1) % BATCH_SIZE == 0 or i == len(df_sample) - 1:
            # Convert the list of processed comments to a DataFrame and save it
            processed_df = pd.DataFrame(processed_comments)
            processed_df.to_csv(OUTPUT_FILE, index=False)
            print(f"✅ Saved progress at {i + 1} comments")

        time.sleep(PROMPT_DELAY)  # Added Delay Between Requests
        print()

    print("🎉 Processing complete! Data saved.")

# Final step: After loop ends, save the processed comments
processed_df = pd.DataFrame(processed_comments)
processed_df.to_csv(OUTPUT_FILE, index=False)
```

```python
import os
import pandas as pd

# Directory where the processed CSV files are located
file_path = "compiled_processed_comments - 12000.csv"

# List to store individual dataframes
dfs = []
     
# Open the file and process line by line
with open(file_path, 'r', encoding='utf-8') as file:
    for line in file:
        if line.startswith("video_id,video_title,comment"):
            continue
        # Strip leading/trailing whitespaces and replace commas separating the fields with semicolons
        line = line.strip()
        
        # Replace the first 3 commas (video_id, video_title, comment) with semicolons
        parts = line.split(',')
        
        if len(parts) >= 4:  # Ensure the line has at least video_id, video_title, comment, label
            # Extract the parts: video_id, video_title, and comment
            video_id = parts[0].replace('"','')
            video_title = parts[1].replace('"','')
            comment_with_commas = ','.join(parts[2:-1])  # Join the comment parts
            label = parts[-1].strip().replace('"','')
            
            # Replace commas inside the comment with semicolons (preserving comment structure)
            comment = comment_with_commas.replace(';', '').replace('"','')
            
            # Create a new line with semicolons separating the fields
            new_line = f"{video_id};{video_title};{comment};{label}"
            
            # Append the processed data to the list of DataFrames
            dfs.append({
                'video_id': video_id,
                'video_title': video_title,
                'comment': comment,
                'label': label
            })

# Create a DataFrame from the processed data
combined_df = pd.DataFrame(dfs)

# Export the combined DataFrame to a single CSV file with semicolons as delimiters
combined_df.to_csv("combined_processed_data.csv", index=False, sep=';')

print("Data successfully combined and exported as 'combined_processed_data.csv'")
```
